```python
from pydantic import BaseModel, Field, ValidationError
```

```python
class User(BaseModel):
  name:str = Field(min_length=3, max_length=50, description= "Name of the user")
  age:int = Field(gt=0, lt=100, description= "Age of the user")
  occupation:str = Field(min_length=5, max_length=50, description= "Job of the user", default=None)
  hobbies:list[str] = Field(default = [])
  email:str = Field(pattern=r'^\S+@\S+\.\S+$', default = None)
```

```python
user = User(name = 'Rakesh', age=11, occupation='Student', hobbies=['Fishing', 'Dancing'])
```

```python
user = User(name = 'Rakesh', age=11, occupation='Student')
```

```python
user = User(name = 'Rakesh', age=11, occupation='Student')
```

```python
user = User(name = 'Rakesh', age=11, occupation='Student', email='abc@gmail.com')
```

```python
name = input("Enter the name :")
age = input("Enter the age :")
occupation = input("Enter the occupation :")
email = input("Enter the email :")
try:
  user = User(name = name, age= age, occupation= occupation, email= email)
except ValidationError as e:
  print(f"There is an error occured : {e}")
```
